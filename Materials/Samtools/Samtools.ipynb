### Using `samtools view`: Converting, Filtering, and Counting Alignments

`samtools view` is a versatile command for working with SAM/BAM/CRAM files. It can convert between formats, filter alignments based on flags or mapping quality, and subset reads by region.

#### Basic Syntax
```bash
samtools view [options] <in.bam> [region]
```

#### Common Options
- **`-b`** : Output **BAM** format (binary). Used when converting from SAM → BAM.  
  Example:  
  ```bash
  samtools view -b input.sam > output.bam
  ```

- **`-h`** : Include the **header** in the output (useful when filtering BAMs but still need the header).  
  Example:  
  ```bash
  samtools view -h -b input.bam chrX > chrX.bam
  ```

- **`-f INT`** : Include only reads with **all** of the specified SAM flag bits set.  
  Example (include only properly paired reads):  
  ```bash
  samtools view -b -f 2 input.bam > proper_pairs.bam
  ```

- **`-F INT`** : Exclude reads with **any** of the specified flag bits set.  
  Example (exclude unmapped reads):  
  ```bash
  samtools view -b -F 4 input.bam > mapped_only.bam
  ```

- **`-q INT`** : Filter reads by **minimum mapping quality**.  
  Example (keep reads with MAPQ ≥ 30):  
  ```bash
  samtools view -b -q 30 input.bam > high_quality.bam
  ```

- **Region filtering (e.g., `chrX`)** : Restrict output to a specific genomic region or chromosome.  
  Example:  
  ```bash
  samtools view -b input.bam chrX > chrX_reads.bam
  ```

- **`-@ INT`** : Use multiple **threads** for faster processing.  
  Example:  
  ```bash
  samtools view -b -@ 8 input.bam > output.bam
  ```

- **`-c`** : **Count** alignments instead of writing them out (suppresses output).  
  Example:  
  ```bash
  samtools view -c -F 4 input.bam
  ```

#### Combined Example
Filter for properly paired reads on chromosome X with MAPQ ≥ 30, using 4 threads, and keep the header:
```bash
samtools view -h -b -f 2 -q 30 -@ 4 input.bam chrX > chrX_filtered.bam
```

This command keeps the header (`-h`), outputs BAM (`-b`), includes only properly paired reads (`-f 2`), filters by mapping quality (`-q 30`), restricts to `chrX`, and uses 4 threads (`-@ 4`).

### Using `samtools sort`: Sorting BAM Files

`samtools sort` sorts alignment files by coordinate (default) or by read name.  
Sorting is an essential step before indexing or many downstream analyses, as most tools (e.g., `samtools index`, `bcftools`, `picard`) require coordinate-sorted BAM files to efficiently access reads by genomic region.

#### Basic Syntax
```bash
samtools sort [options] <in.bam> -o <out.bam>
```

#### Common Options
- **`-@ INT`** : Number of threads to use for sorting (faster processing).
- **`-o FILE`** : Specify the output file name.
- **`-n`** : Sort by **read name** instead of coordinates (useful for some RNA-seq or read-pair analyses).

#### Example
Sort a BAM file by coordinate using 4 threads:
```bash
samtools sort -@ 4 input.bam -o input.sorted.bam
```

#### Why Sorting Matters
- Enables **random access** to specific genomic regions.  
- Required for **indexing** and most **variant calling** workflows.  
- Improves **I/O efficiency** when working with large datasets.

### Using `samtools index`: Indexing BAM Files

`samtools index` creates an index file (`.bai` or `.csi`) for a sorted BAM (or CRAM) file.  
Indexing allows rapid, random access to alignments from specific genomic regions without reading the entire file, essential for visualization and many downstream analyses.

#### Basic Syntax
```bash
samtools index [options] <in.sorted.bam> [out.index]
```

#### Common Options
- **No extra arguments** : Creates a `.bai` index by default.  
- **`-b`** : Force output of a BAM-style index (`.bai`).  
- **`-c`** : Create a CSI index (`.csi`), useful for very large genomes or files >512 Mbp.  
- **`-@ INT`** : Use multiple threads to speed up indexing.

#### Example
Index a coordinate-sorted BAM file using 4 threads:
```bash
samtools index -@ 4 input.sorted.bam
```

This generates an index file:
```
input.sorted.bam.bai
```

#### Why Indexing Matters
- Enables **fast access** to reads aligned to specific regions (e.g., `chrX:100000-200000`).  
- Required for tools like **IGV**, **samtools view [region]**, and most **variant calling** pipelines.  
- Improves **data efficiency** when working with large BAM/CRAM files.

### Using `samtools flagstat`: Alignment Statistics Summary

`samtools flagstat` provides a quick summary of alignment statistics from a SAM, BAM, or CRAM file.  
It counts total reads, mapped reads, properly paired reads, duplicates, and more, making it an easy way to assess data quality and alignment performance.

#### Basic Syntax
```bash
samtools flagstat [options] <in.bam>
```

#### Common Options
- **`-@ INT`** : Use multiple threads for faster processing.  
- **No output file** : Results are printed directly to the terminal (stdout).  
- **Works on any format** : SAM, BAM, or CRAM

#### Example
Generate alignment statistics using 4 threads:
```bash
samtools flagstat -@ 4 input.sorted.bam
```

Typical output includes counts like:
```
123456 + 0 in total (QC-passed reads + QC-failed reads)
120000 + 0 primary
3000 + 0 secondary
456 + 0 supplementary
119000 + 0 mapped (96.4% : N/A)
115000 + 0 paired in sequencing
112000 + 0 properly paired (97.4% : N/A)
```

#### Why `flagstat` Matters
- Provides a **fast quality check** after alignment.  
- Helps identify issues such as low mapping rates or excessive duplicates.  
- Useful for **tracking alignment metrics** in multi-sample workflows.
